# Unified Financial Multimodal RAG (Google Colab T4 - 16GB VRAM)
Notebook hợp nhất toàn bộ 2 nhánh dự án:

- **Hybrid Retriever**: `BAAI/bge-m3` + `SQLite FTS5` + `BAAI/bge-reranker-v2-m3`
